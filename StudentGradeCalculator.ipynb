```python
maths = float(input("Please enter the math subject markes out of 100 : "))
physics = float(input("Please enter the physics subject markes markes out of 100 : "))
chemistry = float(input("Please enter the chemistry subject markes markes out of 100 : "))

percentage = (maths + physics + chemistry)*100/300
print(percentage)
if(percentage >= 90):
    grade = "Grade A"
elif(percentage >= 75):
    grade = "Grade B"
elif(percentage >= 60):
    grade = "Grade C"
elif(percentage >= 40):
    grade = "Grade D"
else:
    grade = "Failed"
if(maths < 33 or physics < 33 or chemistry < 33):
    grade = "Failed"
print(grade)
```
