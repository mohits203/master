**Countdown to a Given Date**

Create a function to calculate remaining days until a given date.
from datetime import datetime

```python
target_date = "31-12-2026"
today = dt.date.today()
target = dt.date.strptime(target_date, "%d-%m-%Y")
remaining = (target - today).days
print("Remaining Days:", remaining)
```
