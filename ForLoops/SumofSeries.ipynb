Write a PHP program to calculate the following series using a loop:
1 + 2 + 3 + ... + n.

Then modify the program to calculate:
1² + 2² + 3² + ... + n²

```python
n = int(input("Please enter a number : "))
sum = int()
for i in range(1, n+1):
    sum = sum + i
print(sum)
```

Then modify the program to calculate:
1² + 2² + 3² + ... + n²

```python
sum = int()
for i in range(1, n+1):
    sum = sum + i*i
print(sum)
```
