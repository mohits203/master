```python
costPrice = float(input("Enter Cost Price : "))
sellingPrice = float(input("Enter Selling Price : "))
quantity = int(input("Enter Quantity : "))

totalCost = costPrice * quantity
totalSelling = sellingPrice * quantity
```

```python
if(totalSelling > totalCost):
    profitOrLoss = totalSelling - totalCost
    profitPer = (profitOrLoss / totalCost) * 100
    if(profitPer >= 20):
        profit = "High Profit"
    else:
        profit = "Normal Profit"
elif(totalSelling < totalCost):
    profitOrLoss = totalCost - totalSelling
    profitPer = (profitOrLoss / totalCost) * 100
    if(profitPer >= 20):
        profit = "High Loss"
    else:
        profit = "Normal Loss" 
else:
    profit = "No Profit, No Loss"

print(f"Profit or Loss is = {profit} and percentage is = {profitPer}")
```
