Write a python program that stores a secret number and repeatedly asks the user to guess it

using a loop. Display "Too High", "Too Low", or "Correct Guess" until the correct number is

entered.

```python
secretNumber = 6
guessNum = int(input("please enter a number : "))
counter = 2
for i in range(1,counter):
    jCounter = 3
    for j in range(1,jCounter):
        if(guessNum < secretNumber):
            print("Too Low")
            guessNum = int(input("please enter a number : "))
            jCounter = 3
        elif(guessNum > secretNumber):
            print("Too High")
            guessNum = int(input("please enter a number : "))
            jCounter = 3
        else:
            print("Correct Guess")
```
