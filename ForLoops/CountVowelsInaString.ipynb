Write a python program to accept a string and count the number of vowels using a loop.
Example: Programming → 3 vowels

```python
stri = input("Please enter a string")
result = int()
for i in stri:
    if(i == "a" or i == "e" or i == "i" or i == "o" or i == "u"):
        result += 1
print(result)
```
