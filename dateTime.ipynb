**strftime()**

Working

Converts a date/time object into a formatted string.

Code Meaning

%Y Year (2026)

%y Year (26)

%m Month

%d Day

%H Hour (24-hour)

%I Hour (12-hour)

%M Minute

%S Second

%A Full weekday

%a Short weekday

%B Full month

%b Short month

%p AM/PM

```python
now = dt.datetime.now()
print(now.strftime("%d-%m-%Y"))
print(now.strftime("%A"))
print(now.strftime("%B"))
print(now.strftime("%I:%M %p"))
```

**strptime()**

Converts a string into a datetime object.

```python
date = dt.datetime.strptime("29-07-2026", "%d-%m-%Y")
print(date)
```

**timedelta()**

Represents the difference between two dates or times.

```python
today = dt.date.today()
next_week = today + dt.timedelta(days=7)
print(next_week)
```

**replace()**

Replaces specific values in a date or datetime object.

```python
d = dt.datetime.now()
new = d.replace(year=2030)
print(new)
```

**weekday()**

**timestamp()**

Returns the Unix timestamp.

```python
timest = dt.datetime.now()
print(timest.timestamp())
```

**fromtimestamp()**

Converts a Unix timestamp into a datetime object.

```python
timestampConvert = dt.datetime.fromtimestamp(1790861803)
print(timestampConvert)
```

**ctime()**

Returns a readable string representation of the date and time.

```python
now = dt.datetime.now()
print(now.ctime())
```

**isoformat()**

Returns the date and time in ISO 8601 format.

```python
now = dt.datetime.now()
print(now.isoformat())
```

**combine()**

Combines a date object and a time object into a datetime object.

```python
d = dt.date(2026, 7, 29)
t = dt.time(10, 30)
cdt = dt.datetime.combine(d, t)
print(cdt)
```

**min and max**

Return the minimum and maximum possible date/time values.

```python
print(dt.datetime.min)
print(dt.datetime.max)
```

**Date Comparison**

Compares two dates.

date1 > date2

date1 < date2

date1 == date2

```python
d1 = dt.date(2026, 7, 29)
d2 = dt.date(2026, 8, 1)
print(d1 < d2)
print(d1 == d2)
```

**Difference Between Two Dates**

Calculates the number of days between two dates.

difference = date2 - date1

```python
d1 = dt.date(2026, 7, 29)
d2 = dt.date(2026, 8, 10)
difference = d2 - d1
print(difference)
print(difference.days)
```
