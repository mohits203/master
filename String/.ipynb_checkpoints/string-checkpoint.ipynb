**startswith()**

The startswith() method checks whether a string starts with a particular character or substring.

It returns either:

True

or

False

```python
text = "Python"
print(text.startswith("Py"))
```

```python
print(text.startswith("Java"))
```

```python
print(name.startswith("K"))
```

**endswith()**

The endswith() method checks whether a string ends with a particular character or substring.

It also returns True or False.

```python
text = "Python"
print(text.endswith("on"))
```

```python
print(text.endswith("kt"))
```

```python
filename = "report.pdf"
print(filename.endswith(".pdf"))
```

**isdigit()**

The isdigit() method checks whether all characters in a string are digits.

It returns:

• True → if all characters are digits

• False → if the string contains non-digit characters

```python
text = "12345"
print(text.isdigit())
```

```python
text = "123abc"
print(text.isdigit())
```

```python
text = "123 45"
print(text.isdigit())
```

**isalpha()**

The isalpha() method checks whether all characters in a string are alphabets/letters.

It returns:

• True → if all characters are letters

• False → if numbers, spaces, or special characters are present

```python
text = "Hello"
print(text.isalpha())
```

```python
text = "Hello123"
print(text.isalpha())
```

```python
text = "Hello World"
#There is a space between "Hello" and "World".
print(text.isalpha())
```

```python
text = "Hello@"
print(text.isalpha())
# @ is not an alphabet.
```

**isalnum()**

The isalnum() method checks whether all characters are either letters or numbers.

The name can be remembered as:

is + alphanumeric

Alphanumeric means:

• A-Z

• a-z

• 0-9

It returns:

• True → letters and/or numbers only

• False → spaces or special characters are present

```python
text = "abc123"
print(text.isalnum())
```

```python
text = "Python"
print(text.isalnum())
```

```python
text = "12345"
print(text.isalnum())
```

```python
text = "abc 123"
print(text.isalnum())
```

```python
text = "abc@123"
print(text.isalnum())
```
