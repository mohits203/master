Because tuples are immutable (they cannot be changed after creation), Python tuples have only two built-in methods built directly into the tuple object. However, you can also use several global built-in functions to work with them.

```python
#Count

myTuple = (1, 3, 7, 8, 7, 5)
print(myTuple.count(7))
```

```python
#index

print(myTuple.index(8))
```

Global Built-in Functions

```python
#sum (sum of all numerical values)

sum(myTuple)
```

```python
#len

len(myTuple)
```

```python
#max

max(myTuple)
```

```python
#min

min(myTuple)
```

```python
#sorted
sorted(myTuple)
```

```python
#convert any list, set into tuple

newList = ["Virat", "Rohit", "KL"]

tupleFromList = tuple(newList)

print(tupleFromList)
print(type(tupleFromList))
```

```python
#any
# We want to check if there are any positive numbers
numbers = (-5, -2, 0, 7, -1)

# Is ANY number greater than 0? 
print(any(x > 0 for x in numbers))  
```

```python
#all
# We want to check if a user filled out all required fields
form_fields = ("Name", "Email", "Age")  # All fields have text

print(all(form_fields))  
# Output: True (because every single field has text)

# If one field is empty ""
broken_form = ("Name", "", "Age")  

print(all(broken_form))  
# Output: False (because the Email field is empty!)
```
